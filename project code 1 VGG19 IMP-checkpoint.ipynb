```python
# Import necessary libraries
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.layers import AveragePooling2D, Dropout, Flatten, Dense, Input
from tensorflow.keras.models import load_model, Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.applications import VGG19
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelBinarizer
from sklearn.metrics import classification_report
from imutils import paths
import matplotlib.pyplot as plt
import numpy as np
import cv2
import os
import pickle

# Define the path to your dataset directory
dataset = "Dataset"  # Change this path to your dataset directory
label = "lb.pickle"

# Define the set of labels you're interested in
LABELS = set(["Fusarium Head Blight", "Healthy Wheat", "Leaf Rust", "Tan Spot"])

# List all image paths in the dataset directory
imagePaths = list(paths.list_images(dataset))
data = []
labels = []

# Load and preprocess images
for imagePath in imagePaths:
    # Extract the label from the image path
    label = imagePath.split(os.path.sep)[-2]
    # Continue if the label is not in the defined LABELS
    if label not in LABELS:
        continue

    # Read and preprocess the image
    image = cv2.imread(imagePath)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image = cv2.resize(image, (224, 224))

    data.append(image)
    labels.append(label)

data = np.array(data)
labels = np.array(labels)

# Check if labels are not empty
if len(labels) == 0:
    raise ValueError("No labels found in the dataset.")

# Initialize the label binarizer
lb = LabelBinarizer()
labels = lb.fit_transform(labels)

# Split the data into training and testing sets
(trainX, testX, trainY, testY) = train_test_split(data, labels,
                                                  test_size=0.25, stratify=labels, random_state=42)

# Set up data augmentation for training
trainAug = ImageDataGenerator(
    rotation_range=30,
    zoom_range=0.15,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.15,
    horizontal_flip=True,
    fill_mode="nearest")

# Set up data augmentation for validation
valAug = ImageDataGenerator()

# Mean subtraction values
mean = np.array([123.68, 116.779, 103.939], dtype="float32")
trainAug.mean = mean
valAug.mean = mean

# Define and compile the model (VGG19)
headmodel = VGG19(weights="imagenet", include_top=False,
                  input_tensor=Input(shape=(224, 224, 3)))  # Using Input here

model = headmodel.output
model = AveragePooling2D(pool_size=(5, 5))(model)
model = Flatten(name="flatten")(model)
model = Dense(512, activation="relu")(model)
model = Dropout(0.4)(model)
model = Dense(len(lb.classes_), activation="softmax")(model)

model = Model(inputs=headmodel.input, outputs=model)

# Freeze layers in the base model
for layer in headmodel.layers:
    layer.trainable = False

# Compile the model
opt = Adam(learning_rate=1e-3)  # Note: use `learning_rate` instead of `lr`
model.compile(loss="categorical_crossentropy", optimizer=opt,
              metrics=["accuracy"])

# Train the model
H = model.fit(
    trainAug.flow(trainX, trainY, batch_size=64),
    steps_per_epoch=len(trainX) // 64,
    validation_data=valAug.flow(testX, testY),
    validation_steps=len(testX) // 64,
    epochs=50)

# Evaluate the model
predictions = model.predict(testX, batch_size=64)

# Ensure that the number of classes in the predictions matches the target names
pred_classes = np.unique(predictions.argmax(axis=1))
target_names = [lb.classes_[i] for i in pred_classes]

# If the number of target names doesn't match the number of classes, adjust them
if len(target_names) != len(lb.classes_):
    print("Warning: Some classes were not predicted. Adjusting target names.")

print(classification_report(testY.argmax(axis=1),
                            predictions.argmax(axis=1), target_names=target_names))

# Plot accuracy and loss
N = 50
plt.plot(np.arange(0, N), H.history['accuracy'], label="Training Accuracy")
plt.plot(np.arange(0, N), H.history['val_accuracy'], label="Validation Accuracy")
plt.title('VGG19 Model Train vs Validation Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(loc='lower right')
plt.show()
plt.savefig("acc_plot.png")

plt.plot(H.history['loss'], label="Training Loss")
plt.plot(H.history['val_loss'], label="Validation Loss")
plt.title('VGG19 Model Train vs Validation Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(loc='upper right')
plt.show()
plt.savefig("loss_plot.png")

# Save the model to disk
model.save("model.h5")

# Save the label binarizer to disk
with open(label, "wb") as f:
    f.write(pickle.dumps(lb))
```

```python
from keras.models import load_model
import numpy as np
import imutils
import cv2
from collections import deque
import pickle

# Load the model and label binarizer
model_path = "model.h5"
input_file = r"C:\Users\shrik\OneDrive\Desktop\minor\leaf3.jpg"
label = "lb.pickle"

moodel = load_model(model_path)
lb = pickle.loads(open(label, "rb").read())

mean = np.array([123.68, 116.779, 103.939][::1], dtype="float32")
Q = deque(maxlen=128)

# Load the input image
frame = cv2.imread(input_file)

if frame is None:
    raise ValueError(f"Image at path '{input_file}' could not be loaded.")

# Preprocess the frame
(H, W) = frame.shape[:2]
output = frame.copy()
frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
frame = cv2.resize(frame, (224, 224)).astype("float32")
frame -= mean

# Make predictions
preds = moodel.predict(np.expand_dims(frame, axis=0))[0]
Q.append(preds)

results = np.array(Q).mean(axis=0)
i = np.argmax(results)
predicted_label = lb.classes_[i]

# Get the prediction confidence (accuracy)
confidence = results[i] * 100  # Convert to percentage

# Display the predicted label and confidence
text = "PREDICTION: {} ({:.2f}%)".format(predicted_label.upper(), confidence)
cv2.putText(output, text, (10, 30), cv2.FONT_HERSHEY_SIMPLEX,
            0.8, (200, 255, 155), 2)

# Display the output image with the prediction and accuracy
cv2.imshow("Output", output)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
